```python
#tarea1 - Metro CDMX
#justificación de datos
#Costo por paso: Asigne un costo unitario c(s, a, s') = 1 a cada tramo entre estaciones contigua, la verdad me hice bolas con las distancias en km y no me salia jeje
#Bajo un esquema de costo uniforme/unitario, la Búsqueda en Anchura (BFS) garantiza encontrar el camino óptimo en número de transbordos y estaciones recorridas.
```

```python
#carga el grafo a la memoria 
class Node:
    def __init__(self, state, parent=None, path_cost=0):#path_cost es el número de estaciones
        self.state = state
        self.parent = parent
        self.path_cost = path_cost  #con costo=1

    def path(self):
        node, path_back = self, []
        while node:
            path_back.append(node.state)
            node = node.parent
        return path_back[::-1]

#se tiene el grafo con las conexiones y $1

metro_cdmx = {
    #dividire por lineas que mas estaciones abarca
    #Línea 1
    'Balderas': {'Hidalgo': 1, 'Salto del Agua': 1, 'Tacubaya': 1},
    'Salto del Agua': {'Balderas': 1, 'Pino Suárez': 1},
    'Pantitlán': {'Chabacano': 1, 'Oceanía': 1, 'Zaragoza': 1},
    #Línea 2
    'Cuatro Caminos': {'Tacuba': 1},
    'Tacuba': {'Cuatro Caminos': 1, 'Hidalgo': 1, 'Tacubaya': 1},
    'Hidalgo': {'Tacuba': 1, 'Bellas Artes': 1, 'Balderas': 1, 'Guerrero': 1},
    'Bellas Artes': {'Hidalgo': 1, 'Pino Suárez': 1},
    'Pino Suárez': {'Bellas Artes': 1, 'Chabacano': 1, 'Salto del Agua': 1},
    'Chabacano': {'Pino Suárez': 1, 'Viaducto': 1, 'Zapata': 1, 'Jamaica': 1},
    'Viaducto': {'Chabacano': 1, 'Taxqueña': 1},
    'Taxqueña': {'Viaducto': 1},
    #como no tenemos lineas completas, coloco los que son destino o transborde sin la linea completa
    'Politécnico': {'La Raza': 1},
    'La Raza': {'Politécnico': 1, 'Guerrero': 1, 'Oceanía': 1},
    'Oceanía': {'La Raza': 1, 'Pantitlán': 1},
    'Guerrero': {'La Raza': 1, 'Hidalgo': 1},
    'Zapata': {'Chabacano': 1, 'Universidad': 1}
}
```

```python
#hacemos los busquedas
from collections import deque
#BFS---------------------------------------------------------------------------------------------
def bfs_metro(graph, start, goal):
    start_node = Node(start)
    if start == goal: return start_node
    frontier = deque([start_node])
    explored = {start}
    while frontier:
        node = frontier.popleft()
        for neighbor, cost in graph.get(node.state, {}).items():
            if neighbor not in explored:
                child = Node(neighbor, parent=node, path_cost=node.path_cost + cost)
                if neighbor == goal: return child
                explored.add(neighbor)
                frontier.append(child)
    return None
#DFS--------------------------------------------------------------------------------------------
def dfs_metro(graph, start, goal):
    start_node = Node(start)
    if start == goal: return start_node
    frontier = [start_node]
    explored = set()
    while frontier:
        node = frontier.pop()
        if node.state == goal: return node
        explored.add(node.state)
        frontier_states = {n.state for n in frontier}
        for neighbor, cost in graph.get(node.state, {}).items():
            if neighbor not in explored and neighbor not in frontier_states:
                child = Node(neighbor, parent=node, path_cost=node.path_cost + cost)
                frontier.append(child)
    return None
```

```python
#ruta 1: Cuatro Caminos -> Pantitlán
print("RUTA 1: ")
origen1, destino1 = 'Cuatro Caminos', 'Pantitlán'

res_bfs1 = bfs_metro(metro_cdmx, origen1, destino1)
res_dfs1 = dfs_metro(metro_cdmx, origen1, destino1)

print(f"Ruta: {origen1} -> {destino1}")
print("BFS:", res_bfs1.path(), f"| Total tramos: {res_bfs1.path_cost}")
print("DFS:", res_dfs1.path(), f"| Total tramos: {res_dfs1.path_cost}")
```

```python
print("RUTA 2: ")
origen2, destino2 = 'Politécnico', 'Taxqueña'

res_bfs2 = bfs_metro(metro_cdmx, origen2, destino2)
res_dfs2 = dfs_metro(metro_cdmx, origen2, destino2)

print(f"Ruta: {origen2} -> {destino2}")
print("BFS:", res_bfs2.path(), f"| Total tramos: {res_bfs2.path_cost}")
print("DFS:", res_dfs2.path(), f"| Total tramos: {res_dfs2.path_cost}")
```

```python
print("RUTA 3: ")
origen3, destino3 = 'Zapata', 'Oceanía'

res_bfs3 = bfs_metro(metro_cdmx, origen3, destino3)
res_dfs3 = dfs_metro(metro_cdmx, origen3, destino3)

print(f"Ruta: {origen3} -> {destino3}")
print("BFS:", res_bfs3.path(), f"| Total tramos: {res_bfs3.path_cost}")
print("DFS:", res_dfs3.path(), f"| Total tramos: {res_dfs3.path_cost}")
```
